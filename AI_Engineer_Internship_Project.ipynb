Q# AI Student Study Assistant

An AI-powered study assistant that helps college students learn and revise using Google Gemini AI.

### Features
- Explain concepts
- Summarize study notes
- Generate quizzes
- Create flashcards

## Project Description

The AI Student Study Assistant is a simple AI-powered application designed to help college students with learning and revision.

It uses the Google Gemini API to generate useful study content based on the student's input.

The application provides explanations, summaries, quizzes, and flashcards through a simple Gradio interface.

## How It Works

1. The student selects a study feature.
2. The student enters a topic or study notes.
3. The application creates a structured prompt.
4. The prompt is sent to the Google Gemini AI model.
5. The generated result is displayed through the Gradio interface.
6. Basic validation and error handling provide a smoother user experience.

## Technologies Used

- Python
- Google Gemini API
- Google GenAI Python SDK
- Gradio
- Google Colab

## Features

### 1. Explain
Explains a topic in simple language with examples and key points.

### 2. Summarize
Converts study notes into concise bullet points for quick revision.

### 3. Quiz
Generates 5 multiple-choice questions with answers and explanations.

### 4. Flashcards
Creates 5 question-and-answer flashcards for exam revision.

```python
from google import genai
from getpass import getpass

api_key = getpass("Enter your Gemini API key: ")

client = genai.Client(api_key=api_key)

print("Gemini connected successfully!")
```

```python
question = input("Enter your study question: ")

if question.strip() == "":
    print("Please enter a question.")
else:
    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=question
    )

    print("\nAI Answer:")
    print(response.text)
```

```python
def study_assistant(feature, content):

    try:
        if feature == "Explain":
            prompt = f"""
You are an AI study assistant for college students.

Explain the topic in simple, easy-to-understand language.

Include:
1. Simple definition
2. Detailed explanation
3. Real-world example
4. Key points
5. Short exam-ready summary

Topic:
{content}
"""

        elif feature == "Summarize":
            prompt = f"""
You are an AI study assistant for college students.

Summarize the following study notes for quick revision.

Include:
1. Main points
2. Important definitions
3. Key concepts
4. Important facts
5. Short revision summary

Use clear bullet points.

Study Notes:
{content}
"""

        elif feature == "Quiz":
            prompt = f"""
You are an AI study assistant.

Create 5 multiple-choice questions about the topic.

For each question:
- Give 4 options (A, B, C, D)
- Clearly show the correct answer
- Give a short explanation of the answer

Topic:
{content}
"""

        elif feature == "Flashcards":
            prompt = f"""
You are an AI study assistant.

Create 5 study flashcards about the topic.

Format each flashcard as:

Flashcard 1
Question:
Answer:

Keep the questions useful for college exam revision.

Topic:
{content}
"""

        response = client.models.generate_content(
            model="gemini-3.6-flash",
            contents=prompt
        )

        return response.text

    except Exception as e:
        return "Sorry, something went wrong. Please try again."
```

```python
print("===== AI STUDY ASSISTANT =====")
print("1. Explain a Concept")
print("2. Summarize Notes")
print("3. Generate a Quiz")

choice = input("\nChoose an option (1-3): ")

if choice == "1":
    feature = "Explain"
    content = input("Enter the topic you want explained: ")

elif choice == "2":
    feature = "Summarize"
    content = input("Paste your study notes: ")

elif choice == "3":
    feature = "Quiz"
    content = input("Enter the topic for the quiz: ")

else:
    print("Invalid choice. Please select 1, 2, or 3.")
    content = ""

if content.strip() != "":
    result = study_assistant(feature, content)

    print("\n===== AI OUTPUT =====")
    print(result)
```

```python
import gradio as gr

def run_assistant(feature, content):
    if content.strip() == "":
        return "Please enter a topic or study notes."

    return study_assistant(feature, content)


app = gr.Interface(
    fn=run_assistant,
    inputs=[
        gr.Dropdown(
            ["Explain", "Summarize", "Quiz", "Flashcards"],
            label="Choose a Study Feature"
        ),
        gr.Textbox(
            label="Enter Topic or Study Notes",
            placeholder="Type your topic or paste your notes here...",
            lines=6
        )
    ],
    outputs=gr.Textbox(label="AI Output"),
    title="🎓 AI Student Study Assistant",
    description="Learn, revise, and test your knowledge with AI."
)

app.launch(share=True)
```
