```python
# Stock Portfolio Tracker

# Hardcoded stock prices
stock_prices = {
    "AAPL": 180,
    "TSLA": 250,
    "GOOGL": 150,
    "MSFT": 420
}

total_investment = 0
portfolio = []

print("📊 Stock Portfolio Tracker")

while True:
    stock_name = input("\nEnter stock name (or type 'done' to finish): ").upper()

    if stock_name == "DONE":
        break

    if stock_name not in stock_prices:
        print("Stock not found!")
        continue

    quantity = int(input("Enter quantity: "))

    price = stock_prices[stock_name]
    investment = price * quantity

    total_investment += investment

    portfolio.append(
        f"{stock_name} - Quantity: {quantity}, Price: {price}, Investment: {investment}"
    )

    print("Investment:", investment)

print("\n----- Portfolio Summary -----")

for item in portfolio:
    print(item)

print("\nTotal investment:", total_investment)

# Save the portfolio to a text file
with open("stock_portfolio.txt", "w") as file:
    file.write("Stock Portfolio Tracker\n")
    file.write("-----------------------\n")

    for item in portfolio:
        file.write(item + "\n")

    file.write(f"\nTotal investment: {total_investment}")

print("\nPortfolio saved successfully as stock_portfolio.txt")
```
